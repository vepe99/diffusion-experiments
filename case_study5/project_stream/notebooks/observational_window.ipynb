```python
import numpy as np
import os

def count_stars_in_window(sim_folder, 
                          n_simulations=10_000, 
                          ra_range_Pal5=None, 
                          dec_range_Pal5=None,
                          ra_range_NGC3201=None,
                            dec_range_NGC3201=None,
                            ra_range_M68=None,
                            dec_range_M68=None, 
                          save_path="star_counts.npy"):
    """
    Counts the number of stars inside the observational window for each simulation.

    Parameters
    ----------
    sim_folder : str
        Path to the folder containing the simulation .npz files.
    n_simulations : int
        Number of simulations to process (default: 10_000).
    ra_range : tuple or None
        (min_ra, max_ra). If None, will not filter on RA.
    dec_range : tuple or None
        (min_dec, max_dec). If None, will not filter on Dec.
    save_path : str
        Path to save the resulting counts array.
    """
    counts_Pal5 = []
    counts_NGC3201 = []
    counts_M68 = []
    for i in tqdm(range(n_simulations)):
        fname = os.path.join(sim_folder, f"simulation_{i}.npz")
        if not os.path.exists(fname):
            print(f"File not found: {fname}")
            continue
        data = np.load(fname)
        if data['j']==0:
            # sim shape: (N_particles, 6), columns 0=ra, 1=dec
            sim = data['sim_data_projected']
            ra = sim[:, 0]
            dec = sim[:, 1]
            mask = np.ones(ra.shape, dtype=bool)
            if ra_range_Pal5 is not None:
                mask &= (ra >= ra_range_Pal5[0]) & (ra <= ra_range_Pal5[1])
            if dec_range_Pal5 is not None:
                mask &= (dec >= dec_range_Pal5[0]) & (dec <= dec_range_Pal5[1])
            counts_Pal5.append(np.sum(mask))
            data.close()
        elif data['j']==1:
             # sim shape: (N_particles, 6), columns 0=ra, 1=dec
            sim = data['sim_data_projected']
            ra = sim[:, 0]
            dec = sim[:, 1]
            mask = np.ones(ra.shape, dtype=bool)
            if ra_range_NGC3201 is not None:
                mask &= (ra >= ra_range_NGC3201[0]) & (ra <= ra_range_NGC3201[1])
            if dec_range_NGC3201 is not None:
                mask &= (dec >= dec_range_NGC3201[0]) & (dec <= dec_range_NGC3201[1])
            counts_NGC3201.append(np.sum(mask))
            data.close()
        elif data['j']==2:
             # sim shape: (N_particles, 6), columns 0=ra, 1=dec
            sim = data['sim_data_projected']
            ra = sim[:, 0]
            dec = sim[:, 1]
            mask = np.ones(ra.shape, dtype=bool)
            if ra_range_M68 is not None:
                mask &= (ra >= ra_range_M68[0]) & (ra <= ra_range_M68[1])
            if dec_range_M68 is not None:
                mask &= (dec >= dec_range_M68[0]) & (dec <= dec_range_M68[1])
            counts_M68.append(np.sum(mask))
            data.close()
    return counts_Pal5, counts_NGC3201, counts_M68

# Example usage:
# Convert to plain floats (degrees)
min_ra_Pal5, max_ra_Pal5 = np.min(track.ra.deg), np.max(track.ra.deg)
min_dec_Pal5, max_dec_Pal5 = np.min(track.dec.deg), np.max(track.dec.deg)

min_ra_NGC3201, max_ra_NGC3201 = np.min(track_ngc3201.ra.deg), np.max(track_ngc3201.ra.deg)
min_dec_NGC3201, max_dec_NGC3201 = np.min(track_ngc3201.dec.deg), np.max(track_ngc3201.dec.deg)

min_ra_M68, max_ra_M68 = np.min(track_m68.ra.deg), np.max(track_m68.ra.deg)
min_dec_M68, max_dec_M68 = np.min(track_m68.dec.deg), np.max(track_m68.dec.deg)

counts_Pal5, counts_NGC3201, counts_M68 = count_stars_in_window(
                                            sim_folder="../data/streams/data/",
                                            n_simulations=70_000,
                                            ra_range_Pal5=(min_ra_Pal5, max_ra_Pal5),
                                            dec_range_Pal5=(min_dec_Pal5, max_dec_Pal5),
                                            ra_range_NGC3201=(min_ra_NGC3201, max_ra_NGC3201),
                                            dec_range_NGC3201=(min_dec_NGC3201, max_dec_NGC3201),
                                            ra_range_M68=(min_ra_M68, max_ra_M68),
                                            dec_range_M68=(min_dec_M68, max_dec_M68),
                                            )
```

```python
fig = plt.figure(figsize=(15,5), tight_layout=True)
ax1 = fig.add_subplot(131)
ax1.hist(counts_Pal5, bins=30, color='blue', alpha=0.7)
ax1.set_title('Star Counts in Pal 5 Observational Window')
ax1.set_xlabel('Number of Stars')
ax1.set_ylabel('Frequency')

ax2 = fig.add_subplot(132)
ax2.hist(counts_NGC3201, bins=30, color='orange', alpha=0.7)
ax2.set_title('Star Counts in NGC 3201 Observational Window')
ax2.set_xlabel('Number of Stars')

ax3 = fig.add_subplot(133)
ax3.hist(counts_M68, bins=30, color='green', alpha=0.7)
ax3.set_title('Star Counts in M68 Observational Window')
ax3.set_xlabel('Number of Stars')
```

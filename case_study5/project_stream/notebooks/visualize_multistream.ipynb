```python
import numpy as np 
import matplotlib.pyplot as plt
data = np.load('../data/streams/data_multistream_galax/simulation_multistream_100.npz')
multistream_test_set = data['sim_data_carthesian']
print('Shape of multistream test set:', multistream_test_set.shape)  # Should be
# multistream_test_set = multistream_test_set.reshape(-1,1000,6)  # Reshape to (n_simulations, n_streams, n_particles, 6)

fig = plt.figure(figsize=(15, 5))
ax1 = fig.add_subplot(131, projection='3d')
ax1.scatter(multistream_test_set[:,0, :, 0], multistream_test_set[:, 0, :, 1], multistream_test_set[:, 0, :, 2], s=1, c='blue', label=data['j'][0, 0, 0])
ax1.legend()
ax1.set_xlim(-20, 20)
ax1.set_ylim(-20, 20)
ax1.set_xlabel('x')
ax1.set_ylabel("y")

ax2 = fig.add_subplot(132, projection='3d')
ax2.scatter(multistream_test_set[:,1, :, 0], multistream_test_set[:, 1, :, 1], multistream_test_set[:, 1, :, 2], s=1, c='green', label=data['j'][0, 1, 0])
ax2.legend()
ax2.set_xlim(-20, 20)
ax2.set_ylim(-20, 20)
ax2.set_xlabel('x')

ax3 = fig.add_subplot(133, projection='3d')
ax3.scatter(multistream_test_set[:,2, :, 0], multistream_test_set[:,2, :, 1], multistream_test_set[:, 2, :, 2], s=1, c='orange', label=data['j'][0, 2, 0])
ax3.legend()
ax3.set_xlim(-20, 20)
ax3.set_ylim(-20, 20)
ax3.set_xlabel('x')
```

```python
j = 1
fig = plt.figure(figsize=(20, 20))
for idx, i in enumerate(range(0, 100)):
    sim_data = multistream_test_set[i, j]

    ax = fig.add_subplot(10, 10, idx+1)
    ax.scatter(sim_data[:, 0], sim_data[:, 1], s=1, label=f"{data['j'][i, j, 0]}")
    # ax.set_xlim(-20, 20)
    # ax.set_ylim(-20, 20)
    ax.legend()
```

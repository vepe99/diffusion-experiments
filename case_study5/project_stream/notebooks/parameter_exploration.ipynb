```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_path = '/export/home/vgiusepp/diffusion-experiments/case_study5/project_stream/data/streams/data_galax_1e6/training_data_local_1000000.npz'
# data_path = '/export/home/vgiusepp/diffusion-experiments/case_study5/project_stream/data/streams/data_multistream_galax/simulation_multistream_1000.npz'
data = np.load(data_path)
data = {k: data[k][:10_000] for k in data.files}
params = ["m_Triaxial_halo", "r_Triaxial_halo", "q2_Triaxial_halo", "rho_thin_disk", "hr_thin_disk", "hz_thin_disk", "rho_thick_disk"]

param_values = {param: data[param] for param in params}
fig = plt.figure(figsize=(20, 5))
for k in param_values.keys():
    ax = fig.add_subplot(2, 4, params.index(k)+1)
    ax.hist(param_values[k], bins=50)
    ax.set_title(k)
plt.tight_layout()
plt.show()
```
